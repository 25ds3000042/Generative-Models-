```python
import tensorflow as tf
import numpy as np

# Our sentence
sentence = "when will we eat the breakfast"

# Split sentence into words
words = sentence.split()

print(words)
```

```python
vocab = sorted(set(words))

word_to_id = {word: i for i, word in enumerate(vocab)}
id_to_word = {i: word for word, i in word_to_id.items()}

print(word_to_id)
```

```python
X = []
y = []

for i in range(len(words) - 1):
    X.append([word_to_id[word] for word in words[:i+1]])
    y.append(word_to_id[words[i+1]])

print("X =", X)
print("y =", y)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences

X = pad_sequences(X, padding="pre")

X = np.array(X)
y = np.array(y)

print(X)
print(y)
```

```python
model = tf.keras.Sequential([
    tf.keras.layers.Embedding(
        input_dim=len(vocab),
        output_dim=16
    ),

    tf.keras.layers.SimpleRNN(32),

    tf.keras.layers.Dense(len(vocab), activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python
history = model.fit(
    X,
    y,
    epochs=500,
    verbose=0
)

print("Training complete!")
```

```python
def predict_next_word(text):

    input_words = text.split()

    input_ids = [
        word_to_id[word]
        for word in input_words
        if word in word_to_id
    ]

    input_ids = pad_sequences(
        [input_ids],
        maxlen=X.shape[1],
        padding="pre"
    )

    prediction = model.predict(input_ids, verbose=0)

    predicted_id = np.argmax(prediction[0])

    return id_to_word[predicted_id]
```

```python
print(predict_next_word("when"))
```

```python
print(predict_next_word("when will"))
```
